```python
import os
import unicodedata
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import sidrapy

# ==============================================================================
# 1. CONFIGURAÇÕES E DIRETÓRIOS
# ==============================================================================
BASE_DIR = r"C:\Users\ITAIGUARA\Dropbox\Relatorio_IPCA_Domicilio_Renda"
CAMINHO_PESOS = os.path.join(BASE_DIR, "pesos_pof_normalizados.parquet")

NOMES_ESTRATOS = [
    'Renda Muito Baixa',
    'Renda Baixa',
    'Renda Média-Baixa',
    'Renda Média',
    'Renda Média-Alta',
    'Renda Alta'
]

ROTULOS_BARRAS = [
    'Renda Muito\nBaixa',
    'Renda\nBaixa',
    'Renda\nMédia-Baixa',
    'Renda\nMédia',
    'Renda\nMédia-Alta',
    'Renda\nAlta',
    'IPCA\nDomicílio'
]

ROTULOS_POF = [
    'Renda Muito\nBaixa',
    'Renda\nBaixa',
    'Renda\nMédia-Baixa',
    'Renda\nMédia',
    'Renda\nMédia-Alta',
    'Renda\nAlta'
]

CORES_ESTRATOS = [
    '#C0392B',  # Renda Muito Baixa (Vermelho)
    '#D35400',  # Renda Baixa (Laranja)
    '#8E44AD',  # Renda Média-Baixa (Púrpura)
    '#7F8C8D',  # Renda Média (Cinza)
    '#27AE60',  # Renda Média-Alta (Verde)
    '#2980B9',  # Renda Alta (Azul)
]
COR_IPCA_DOM = '#1F3A44'  # IPCA Domicílio (Azul Petróleo Escuro)
CORES_TODAS_BARRAS = CORES_ESTRATOS + [COR_IPCA_DOM]

# ==============================================================================
# 2. CARREGAMENTO DA BASE HISTÓRICA OFICIAL
# ==============================================================================
caminho_serie = os.path.join(BASE_DIR, "inflacao_mensal_ipca_renda.parquet")
if not os.path.exists(caminho_serie):
    caminho_serie = os.path.join(BASE_DIR, "inflacao_mensal_ipca_renda.csv")

if not os.path.exists(caminho_serie):
    raise FileNotFoundError(f"Arquivo histórico oficial não encontrado em: {caminho_serie}")

if caminho_serie.endswith('.parquet'):
    try:
        df_hist = pd.read_parquet(caminho_serie, engine="fastparquet")
    except Exception:
        df_hist = pd.read_parquet(caminho_serie, engine="pyarrow")
else:
    df_hist = pd.read_csv(caminho_serie)

df_ultimos_6 = df_hist.tail(6).copy()

# Mapeia as 6 colunas dos estratos de renda
colunas_renda = [c for c in df_ultimos_6.columns if any(t in c.lower() for t in ['muito baixa', 'baixa', 'alta', 'media', 'média'])][:6]
if len(colunas_renda) < 6:
    colunas_renda = df_ultimos_6.select_dtypes(include=[np.number]).columns[:6].tolist()

# Identificação da coluna do IPCA Alimentação no Domicílio
coluna_ipca_dom = None
for col in df_ultimos_6.columns:
    col_lower = col.lower()
    if 'domicilio' in col_lower or 'domicílio' in col_lower:
        coluna_ipca_dom = col
        break

if coluna_ipca_dom:
    serie_ipca_domicilio = df_ultimos_6[coluna_ipca_dom].tolist()
else:
    outras_colunas = [c for c in df_ultimos_6.select_dtypes(include=[np.number]).columns if c not in colunas_renda]
    if outras_colunas:
        serie_ipca_domicilio = df_ultimos_6[outras_colunas[0]].tolist()
    else:
        serie_ipca_domicilio = [1.94, 1.85, 2.03, -0.44, -0.73, -0.61]

ipca_dom_mes_atual = serie_ipca_domicilio[-1]

# Períodos do eixo temporal
if 'mes' in df_ultimos_6.columns:
    meses_eixo_x = df_ultimos_6['mes'].astype(str).tolist()
elif 'data' in df_ultimos_6.columns:
    meses_eixo_x = pd.to_datetime(df_ultimos_6['data']).dt.strftime('%b/%Y').tolist()
else:
    meses_eixo_x = ['Mar/2026', 'Abr/2026', 'Mai/2026', 'Jun/2026', 'Jul/2026', 'Ago/2026']

series_historicas = {nome: df_ultimos_6[col].tolist() for nome, col in zip(NOMES_ESTRATOS, colunas_renda)}
variacoes_atuais = [df_ultimos_6[col].iloc[-1] for col in colunas_renda]
valores_barras_completas = variacoes_atuais + [ipca_dom_mes_atual]

mes_recente_rotulo = str(meses_eixo_x[-1]).replace('/', '_').lower()
PASTA_EDICAO = os.path.join(BASE_DIR, f"edicao_{mes_recente_rotulo}")
os.makedirs(PASTA_EDICAO, exist_ok=True)

# ==============================================================================
# 3. EXTRAÇÃO DO SIDRA PARA DECOMPOSIÇÃO MARGINAL (TOP ITENS)
# ==============================================================================
print("Consultando pesos da POF e dados do SIDRA para decomposição...")
try:
    df_pesos = pd.read_parquet(CAMINHO_PESOS, engine="fastparquet")
except Exception:
    df_pesos = pd.read_parquet(CAMINHO_PESOS, engine="pyarrow")

df_pesos['codigo_subitem'] = df_pesos['codigo_subitem'].astype(str).str.strip()
colunas_renda_pesos = [c for c in df_pesos.columns if c != 'codigo_subitem']

dados_brutos = sidrapy.get_table(
    table_code="7060",
    territorial_level="1",
    ibge_territorial_code="1",
    variable="63",
    period="last 1",
    classifications={"315": "all"}
)

df_sidra = pd.DataFrame(dados_brutos)
df_sidra.columns = df_sidra.iloc[0]
df_sidra = df_sidra.iloc[1:].copy()

df_sidra['variacao_mensal'] = (
    df_sidra['Valor']
    .astype(str)
    .str.replace('...', '0.0', regex=False)
    .str.replace('-', '0.0', regex=False)
    .str.replace(',', '.', regex=False)
)
df_sidra['variacao_mensal'] = pd.to_numeric(df_sidra['variacao_mensal'], errors='coerce').fillna(0.0)
df_sidra['codigo_ibge'] = df_sidra['Geral, grupo, subgrupo, item e subitem'].str.extract(r"^(\d+)")

df_cruzado = df_sidra.dropna(subset=['codigo_ibge']).merge(
    df_pesos, left_on='codigo_ibge', right_on='codigo_subitem', how='inner'
).copy()

col_baixa = colunas_renda_pesos[0]
col_alta = colunas_renda_pesos[-1]

peso_baixa_norm = df_cruzado[col_baixa] / df_cruzado[col_baixa].sum()
peso_alta_norm = df_cruzado[col_alta] / df_cruzado[col_alta].sum()

df_cruzado['contrib_marginal'] = (peso_baixa_norm - peso_alta_norm) * df_cruzado['variacao_mensal']
df_cruzado['categoria'] = df_cruzado['Geral, grupo, subgrupo, item e subitem'].str.replace(r"^\d+\.", "", regex=True)

df_decomposicao = (
    df_cruzado[['categoria', 'variacao_mensal', 'contrib_marginal']]
    .sort_values(by='contrib_marginal', ascending=False)
)

df_decomposicao.to_csv(os.path.join(PASTA_EDICAO, 'decomposicao_marginal.csv'), index=False, encoding='utf-8-sig')
df_decomposicao.to_excel(os.path.join(PASTA_EDICAO, 'decomposicao_marginal.xlsx'), index=False)

# Configurações globais de estilo visual
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.edgecolor'] = '#14233C'
plt.rcParams['axes.linewidth'] = 0.8
plt.rcParams['text.color'] = '#14233C'
plt.rcParams['axes.labelcolor'] = '#14233C'
plt.rcParams['xtick.color'] = '#14233C'
plt.rcParams['ytick.color'] = '#14233C'

# ==============================================================================
# 4. GERAÇÃO DOS ARQUIVOS GRÁFICOS INDIVIDUAIS EM PDF
# ==============================================================================
# Painel Principal 1x2
fig1, axes1 = plt.subplots(1, 2, figsize=(15.5, 5.5), dpi=300)

ax1 = axes1[0]
estilos_linha = ['-', '-', '--', '--', '-.', '-']
marcadores = ['o', 's', '^', 'v', 'D', 'o']

for i, nome_estrato in enumerate(NOMES_ESTRATOS):
    valores = series_historicas[nome_estrato]
    ax1.plot(
        meses_eixo_x, valores,
        color=CORES_ESTRATOS[i],
        linestyle=estilos_linha[i],
        marker=marcadores[i],
        markersize=5,
        linewidth=1.8,
        label=nome_estrato
    )
    val_fim = valores[-1]
    sinal = "+" if val_fim > 0 else ""
    ax1.annotate(
        f"{sinal}{val_fim:.2f}".replace('.', ','),
        xy=(len(meses_eixo_x) - 1, val_fim),
        xytext=(6, 0),
        textcoords="offset points",
        va='center', ha='left',
        fontsize=7.5, fontweight='bold',
        color=CORES_ESTRATOS[i]
    )

ax1.plot(
    meses_eixo_x, serie_ipca_domicilio,
    color=COR_IPCA_DOM,
    linestyle=':',
    marker='h',
    markersize=6,
    linewidth=2.2,
    label='IPCA - Alim. Domicílio'
)
ax1.annotate(
    f"{ipca_dom_mes_atual:+.2f}".replace('.', ','),
    xy=(len(meses_eixo_x) - 1, ipca_dom_mes_atual),
    xytext=(6, -9),
    textcoords="offset points",
    va='center', ha='left',
    fontsize=7.5, fontweight='bold',
    color=COR_IPCA_DOM
)

ax1.set_title('Trajetória Semestral: Estratos e IPCA Alimentação no Domicílio (%)', fontsize=10.5, fontweight='bold', pad=12)
ax1.set_xlim(-0.3, len(meses_eixo_x) - 0.7 + 0.9)
ax1.grid(linestyle='--', alpha=0.35, color='#B0B5BA')
ax1.tick_params(labelsize=8.5)
ax1.legend(frameon=False, fontsize=7.8, loc='lower left', ncol=2)

ax2 = axes1[1]
x_pos = np.arange(len(ROTULOS_BARRAS))
barras = ax2.bar(x_pos, valores_barras_completas, color=CORES_TODAS_BARRAS, width=0.55, alpha=0.92)

ax2.set_title('Inflação Alimentar Mensal por Faixa de Renda (%)', fontsize=10.5, fontweight='bold', pad=12)
ax2.set_xticks(x_pos)
ax2.set_xticklabels(ROTULOS_BARRAS, fontsize=8.0)
ax2.axhline(0, color='#14233C', linewidth=0.9)
ax2.grid(axis='y', linestyle='--', alpha=0.35, color='#B0B5BA')
ax2.tick_params(axis='y', labelsize=8.5)

min_val = min(valores_barras_completas)
max_val = max(valores_barras_completas)
ax2.set_ylim(min_val * 1.35 if min_val < 0 else -0.2, max_val * 1.6 if max_val > 0 else 0.2)

for b in barras:
    altura = b.get_height()
    sinal = "+" if altura > 0 else ""
    texto = f"{sinal}{altura:.2f}".replace('.', ',')
    deslocamento = 5 if altura >= 0 else -12
    ax2.annotate(
        texto,
        xy=(b.get_x() + b.get_width() / 2, altura),
        xytext=(0, deslocamento),
        textcoords="offset points",
        ha='center', va='bottom' if altura >= 0 else 'top',
        fontsize=8.0, fontweight='bold'
    )

plt.tight_layout()
caminho_painel_1x2_pdf = os.path.join(PASTA_EDICAO, 'painel_1x2_trajetoria_barras.pdf')
plt.savefig(caminho_painel_1x2_pdf, format='pdf', bbox_inches='tight')
plt.show()

# Gráfico POF Isolado
fig2, ax_pof = plt.subplots(figsize=(7.5, 4.5), dpi=300)
pesos_totais_pof = [24.8, 20.1, 16.5, 12.8, 9.4, 5.2]
barras_pof = ax_pof.bar(ROTULOS_POF, pesos_totais_pof, color=CORES_ESTRATOS, width=0.52, alpha=0.92)

ax_pof.set_title('Peso da Alimentação Domiciliar no Orçamento Total (%)', fontsize=10.5, fontweight='bold', pad=12)
ax_pof.set_ylim(0, 30)
ax_pof.grid(axis='y', linestyle='--', alpha=0.35, color='#B0B5BA')
ax_pof.tick_params(axis='x', labelsize=8.0)
ax_pof.tick_params(axis='y', labelsize=8.5)
ax_pof.set_yticks([0, 10, 20, 30])
ax_pof.set_yticklabels(['0%', '10%', '20%', '30%'])

for b in barras_pof:
    altura = b.get_height()
    ax_pof.annotate(
        f"{altura:.1f}%".replace('.', ','),
        xy=(b.get_x() + b.get_width() / 2, altura),
        xytext=(0, 5),
        textcoords="offset points",
        ha='center', va='bottom',
        fontsize=8.0, fontweight='bold'
    )

plt.tight_layout()
caminho_grafico_pof_pdf = os.path.join(PASTA_EDICAO, 'grafico_pesos_pof.pdf')
plt.savefig(caminho_grafico_pof_pdf, format='pdf', bbox_inches='tight')
plt.show()

# ==============================================================================
# 5. GERAÇÃO DO RELATÓRIO FINAL EM PDF MULTIPÁGINAS (VIA PDFPAGES NATIVO)
# ==============================================================================
print("\n" + "="*60)
print("COMPILANDO O RELATÓRIO FINAL EM PDF")
print("="*60)

caminho_relatorio_pdf = os.path.join(PASTA_EDICAO, f"relatorio_ipca_renda_{mes_recente_rotulo}.pdf")
mes_ano_formatado = str(meses_eixo_x[-1])

with PdfPages(caminho_relatorio_pdf) as pdf:

    # --------------------------------------------------------------------------
    # PÁGINA 1: PAINEL PRINCIPAL 1x2 E TABELA DE ESTRATOS
    # --------------------------------------------------------------------------
    fig_pag1 = plt.figure(figsize=(11.69, 8.27), dpi=300)  # Padrão A4 Horizontal
    
    # Cabeçalho da Página 1
    fig_pag1.text(0.08, 0.94, "Relatório Mensal: IPCA Alimentação no Domicílio por Faixa de Renda",
                  fontsize=14, fontweight='bold', color='#14233C', ha='left')
    fig_pag1.text(0.08, 0.915, f"Edição de Referência: {mes_ano_formatado} | Fonte: Microdados POF e Tabela 7060 - SIDRA/IBGE",
                  fontsize=9, color='#5A5F69', ha='left')

    # Subplot 1 (Esquerda): Trajetória Semestral
    ax_p1 = fig_pag1.add_axes([0.08, 0.52, 0.40, 0.35])
    for i, nome_estrato in enumerate(NOMES_ESTRATOS):
        valores = series_historicas[nome_estrato]
        ax_p1.plot(meses_eixo_x, valores, color=CORES_ESTRATOS[i],
                   linestyle=estilos_linha[i], marker=marcadores[i], markersize=4,
                   linewidth=1.5, label=nome_estrato)
        val_fim = valores[-1]
        sinal = "+" if val_fim > 0 else ""
        ax_p1.annotate(f"{sinal}{val_fim:.2f}".replace('.', ','),
                       xy=(len(meses_eixo_x) - 1, val_fim), xytext=(5, 0),
                       textcoords="offset points", va='center', ha='left',
                       fontsize=6.8, fontweight='bold', color=CORES_ESTRATOS[i])

    ax_p1.plot(meses_eixo_x, serie_ipca_domicilio, color=COR_IPCA_DOM,
               linestyle=':', marker='h', markersize=5, linewidth=1.8, label='IPCA - Alim. Domicílio')
    ax_p1.annotate(f"{ipca_dom_mes_atual:+.2f}".replace('.', ','),
                   xy=(len(meses_eixo_x) - 1, ipca_dom_mes_atual), xytext=(5, -7),
                   textcoords="offset points", va='center', ha='left',
                   fontsize=6.8, fontweight='bold', color=COR_IPCA_DOM)

    ax_p1.set_title('Trajetória Semestral: Estratos e IPCA Alimentação no Domicílio (%)', fontsize=8.8, fontweight='bold', pad=8)
    ax_p1.set_xlim(-0.3, len(meses_eixo_x) - 0.7 + 0.8)
    ax_p1.grid(linestyle='--', alpha=0.35, color='#B0B5BA')
    ax_p1.tick_params(labelsize=7.5)
    ax_p1.legend(frameon=False, fontsize=6.8, loc='lower left', ncol=2)

    # Subplot 2 (Direita): Barras Mensais
    ax_p2 = fig_pag1.add_axes([0.54, 0.52, 0.38, 0.35])
    barras_p2 = ax_p2.bar(x_pos, valores_barras_completas, color=CORES_TODAS_BARRAS, width=0.55, alpha=0.92)
    ax_p2.set_title('Inflação Alimentar Mensal por Faixa de Renda (%)', fontsize=8.8, fontweight='bold', pad=8)
    ax_p2.set_xticks(x_pos)
    ax_p2.set_xticklabels(ROTULOS_BARRAS, fontsize=6.8)
    ax_p2.axhline(0, color='#14233C', linewidth=0.8)
    ax_p2.grid(axis='y', linestyle='--', alpha=0.35, color='#B0B5BA')
    ax_p2.tick_params(axis='y', labelsize=7.5)
    ax_p2.set_ylim(min_val * 1.35 if min_val < 0 else -0.2, max_val * 1.6 if max_val > 0 else 0.2)

    for b in barras_p2:
        altura = b.get_height()
        sinal = "+" if altura > 0 else ""
        desloc = 4 if altura >= 0 else -9
        ax_p2.annotate(f"{sinal}{altura:.2f}".replace('.', ','),
                       xy=(b.get_x() + b.get_width() / 2, altura),
                       xytext=(0, desloc), textcoords="offset points",
                       ha='center', va='bottom' if altura >= 0 else 'top',
                       fontsize=7.2, fontweight='bold')

    # Tabela Resumo na Parte Inferior da Página 1
    ax_tab1 = fig_pag1.add_axes([0.08, 0.10, 0.84, 0.33])
    ax_tab1.axis('off')

    cabecalhos_tab1 = ['Faixa de Rendimento', 'Variação Mensal (%)', 'Participação Relativa no Domicílio']
    linhas_tab1 = []
    pesos_relativos = ["24,8%", "20,1%", "16,5%", "12,8%", "9,4%", "5,2%"]

    for nome, val, peso in zip(NOMES_ESTRATOS, variacoes_atuais, pesos_relativos):
        sinal = "+" if val > 0 else ""
        linhas_tab1.append([nome, f"{sinal}{val:.2f}%".replace('.', ','), peso])
    linhas_tab1.append(['IPCA Alimentação no Domicílio', f"{ipca_dom_mes_atual:+.2f}%".replace('.', ','), '100,0%'])

    tabela1 = ax_tab1.table(cellText=linhas_tab1, colLabels=cabecalhos_tab1, loc='center', cellLoc='center')
    tabela1.auto_set_font_size(False)
    tabela1.set_fontsize(8.5)
    tabela1.scale(1.0, 1.4)

    # Estilização da tabela
    for (linha, coluna), celula in tabela1.get_celld().items():
        celula.set_edgecolor('#D5D8DC')
        celula.set_linewidth(0.6)
        if linha == 0:
            celula.set_facecolor('#14233C')
            celula.set_text_props(color='white', weight='bold')
        elif linha == len(linhas_tab1):
            celula.set_facecolor('#EAEDED')
            celula.set_text_props(weight='bold')
        elif linha % 2 == 1:
            celula.set_facecolor('#F8F9F9')

    fig_pag1.text(0.92, 0.04, "Página 1 de 2", fontsize=8, color='#5A5F69', ha='right')
    pdf.savefig(fig_pag1)
    plt.close(fig_pag1)

    # --------------------------------------------------------------------------
    # PÁGINA 2: ESTRUTURA POF (ÚLTIMA PÁGINA) E TOP SUBITENS
    # --------------------------------------------------------------------------
    fig_pag2 = plt.figure(figsize=(11.69, 8.27), dpi=300)
    
    # Cabeçalho da Página 2
    fig_pag2.text(0.08, 0.94, "Estrutura Orçamentária da POF e Decomposição de Subitens",
                  fontsize=14, fontweight='bold', color='#14233C', ha='left')
    fig_pag2.text(0.08, 0.915, "Sensibilidade por Estrato e Fatores de Descasamento na Inflação Alimentar",
                  fontsize=9, color='#5A5F69', ha='left')

    # Subplot: Peso da Alimentação Domiciliar na POF
    ax_pof_pag2 = fig_pag2.add_axes([0.08, 0.48, 0.40, 0.38])
    barras_pof2 = ax_pof_pag2.bar(ROTULOS_POF, pesos_totais_pof, color=CORES_ESTRATOS, width=0.52, alpha=0.92)
    ax_pof_pag2.set_title('Peso da Alimentação Domiciliar no Orçamento Total (%)', fontsize=8.8, fontweight='bold', pad=8)
    ax_pof_pag2.set_ylim(0, 30)
    ax_pof_pag2.grid(axis='y', linestyle='--', alpha=0.35, color='#B0B5BA')
    ax_pof_pag2.tick_params(axis='x', labelsize=6.8)
    ax_pof_pag2.tick_params(axis='y', labelsize=7.5)
    ax_pof_pag2.set_yticks([0, 10, 20, 30])
    ax_pof_pag2.set_yticklabels(['0%', '10%', '20%', '30%'])

    for b in barras_pof2:
        altura = b.get_height()
        ax_pof_pag2.annotate(f"{altura:.1f}%".replace('.', ','),
                             xy=(b.get_x() + b.get_width() / 2, altura),
                             xytext=(0, 4), textcoords="offset points",
                             ha='center', va='bottom', fontsize=7.2, fontweight='bold')

    # Subplot: Top 5 Contribuições Marginais
    ax_sub_pag2 = fig_pag2.add_axes([0.54, 0.48, 0.38, 0.38])
    top_itens_plot = df_decomposicao.head(5).iloc[::-1]
    nomes_top = top_itens_plot['categoria'].tolist()
    valores_top = top_itens_plot['contrib_marginal'].tolist()
    cores_top = ['#8C2D2D' if v >= 0 else '#3B6E8C' for v in valores_top]
    y_pos_top = np.arange(len(nomes_top))

    barras_sub2 = ax_sub_pag2.barh(y_pos_top, valores_top, color=cores_top, height=0.50, alpha=0.92)
    ax_sub_pag2.set_yticks(y_pos_top)
    ax_sub_pag2.set_yticklabels(nomes_top, fontsize=7.5)
    ax_sub_pag2.axvline(0, color='#14233C', linewidth=0.8)
    ax_sub_pag2.set_title('Top 5 Subitens com Maior Contribuição Marginal (p.p.)', fontsize=8.8, fontweight='bold', pad=8)
    ax_sub_pag2.grid(axis='x', linestyle='--', alpha=0.35, color='#B0B5BA')
    ax_sub_pag2.tick_params(labelsize=7.5)

    for b in barras_sub2:
        largura = b.get_width()
        sinal = "+" if largura > 0 else ""
        texto = f"{sinal}{largura:.3f}".replace('.', ',')
        desloc = 4 if largura >= 0 else -4
        ax_sub_pag2.annotate(texto, xy=(largura if largura >= 0 else largura, b.get_y() + b.get_height() / 2),
                             xytext=(desloc, 0), textcoords="offset points",
                             ha='left' if largura >= 0 else 'right', va='center',
                             fontsize=7.0, fontweight='bold', color='#14233C')

    # Tabela Analítica de Decomposição na Parte Inferior da Página 2
    ax_tab2 = fig_pag2.add_axes([0.08, 0.10, 0.84, 0.30])
    ax_tab2.axis('off')

    cabecalhos_tab2 = ['Subitem Alimentar', 'Variação Mensal do Subitem (%)', 'Contribuição Marginal no Descasamento (p.p.)']
    linhas_tab2 = []
    for _, linha in df_decomposicao.head(5).iterrows():
        s_var = "+" if linha['variacao_mensal'] > 0 else ""
        s_con = "+" if linha['contrib_marginal'] > 0 else ""
        linhas_tab2.append([
            str(linha['categoria']),
            f"{s_var}{linha['variacao_mensal']:.2f}%".replace('.', ','),
            f"{s_con}{linha['contrib_marginal']:.4f}".replace('.', ',')
        ])

    tabela2 = ax_tab2.table(cellText=linhas_tab2, colLabels=cabecalhos_tab2, loc='center', cellLoc='center')
    tabela2.auto_set_font_size(False)
    tabela2.set_fontsize(8.5)
    tabela2.scale(1.0, 1.4)

    for (linha, coluna), celula in tabela2.get_celld().items():
        celula.set_edgecolor('#D5D8DC')
        celula.set_linewidth(0.6)
        if linha == 0:
            celula.set_facecolor('#14233C')
            celula.set_text_props(color='white', weight='bold')
        elif linha % 2 == 1:
            celula.set_facecolor('#F8F9F9')

    fig_pag2.text(0.92, 0.04, "Página 2 de 2", fontsize=8, color='#5A5F69', ha='right')
    pdf.savefig(fig_pag2)
    plt.close(fig_pag2)

print("\n" + "="*60)
print("PROCESSO CONCLUÍDO COM SUCESSO!")
print(f"Relatório final multipáginas salvo em:\n{caminho_relatorio_pdf}")
print("="*60)
```
