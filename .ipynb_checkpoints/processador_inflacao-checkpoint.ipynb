```python
import os
import pandas as pd
import numpy as np
import sidrapy

BASE_DIR = r"C:\Users\ITAIGUARA\Dropbox\Relatorio_IPCA_Domicilio_Renda"

# Matriz de pesos relativos da alimentação no domicílio da POF (Normalizada para soma = 1.0 por faixa)
PESOS_POF_NORMALIZADOS = {
    'Cereais e leguminosas':      [0.108, 0.082, 0.065, 0.048, 0.035, 0.021],
    'Farinhas, féculas e massas': [0.062, 0.048, 0.038, 0.029, 0.021, 0.014],
    'Tubérculos e raízes':        [0.041, 0.035, 0.031, 0.026, 0.022, 0.016],
    'Açúcares e derivados':       [0.045, 0.038, 0.030, 0.024, 0.018, 0.012],
    'Hortaliças e verduras':      [0.051, 0.055, 0.062, 0.068, 0.072, 0.078],
    'Frutas':                     [0.058, 0.069, 0.082, 0.095, 0.112, 0.135],
    'Carnes e aves':              [0.245, 0.231, 0.215, 0.198, 0.175, 0.142],
    'Pescados':                   [0.028, 0.026, 0.025, 0.024, 0.025, 0.028],
    'Leites e derivados':         [0.135, 0.142, 0.148, 0.152, 0.155, 0.158],
    'Panificados':                [0.112, 0.105, 0.098, 0.091, 0.082, 0.071],
    'Óleos e gorduras':           [0.038, 0.031, 0.025, 0.020, 0.016, 0.011],
    'Bebidas e infusões':         [0.045, 0.052, 0.061, 0.075, 0.088, 0.114],
    'Outros alimentos':           [0.032, 0.086, 0.120, 0.150, 0.180, 0.200],
}

ESTRATOS_NOMES = [
    'Até 3 SM',
    '3 a 6 SM',
    '6 a 10 SM',
    '10 a 15 SM',
    '15 a 25 SM',
    '> 25 SM'
]

# Códigos oficiais dos subgrupos na classificação 315 do SIDRA (Tabela 7060)
CODIGOS_SUBGRUPOS_SIDRA = {
    '7172': 'Cereais e leguminosas',
    '7173': 'Farinhas, féculas e massas',
    '7174': 'Tubérculos e raízes',
    '7175': 'Açúcares e derivados',
    '7176': 'Hortaliças e verduras',
    '7177': 'Frutas',
    '7178': 'Carnes e aves',
    '7179': 'Pescados',
    '7182': 'Leites e derivados',
    '7183': 'Panificados',
    '7184': 'Óleos e gorduras',
    '7185': 'Bebidas e infusões',
}

def extrair_dados_completos_sidra(periodo="last 6"):
    """Consulta os dados oficiais do IPCA na Tabela 7060 do SIDRA."""
    print(f"Consultando Tabela 7060 do SIDRA para o período: {periodo}...")
    
    # Consulta Índice Geral (7169), Alimentação Geral (7170) e os subgrupos
    codigos_busca = ['7169', '7170', '7171'] + list(CODIGOS_SUBGRUPOS_SIDRA.keys())
    codigos_str = ",".join(codigos_busca)

    try:
        dados_brutos = sidrapy.get_table(
            table_code="7060",
            territorial_level="1",
            ibge_territorial_code="all",
            variable="63", # Variação mensal (%)
            period=periodo,
            classifications={"315": codigos_str}
        )
    except Exception as erro:
        raise RuntimeError(f"Erro na conexão com SIDRA: {erro}")

    df = pd.DataFrame(dados_brutos)
    df.columns = df.iloc[0]
    df = df.iloc[1:].copy()

    df_selecionado = df[[
        'Mês (Código)', 
        'Mês', 
        'Geral, grupo, subgrupo, item e subitem (Código)',
        'Geral, grupo, subgrupo, item e subitem', 
        'Valor'
    ]].copy()
    
    df_selecionado.columns = ['mes_codigo', 'mes_nome', 'codigo_item', 'nome_item', 'variacao_mensal']

    # Tratamento numérico
    df_selecionado['variacao_mensal'] = (
        df_selecionado['variacao_mensal']
        .astype(str)
        .str.replace('...', '0.0', regex=False)
        .str.replace('-', '0.0', regex=False)
        .str.replace(',', '.', regex=False)
    )
    df_selecionado['variacao_mensal'] = pd.to_numeric(df_selecionado['variacao_mensal'], errors='coerce').fillna(0.0)
    
    return df_selecionado

def processar_calculos_relatorio(periodo="last 6"):
    """
    Executa o cruzamento da POF com os dados do SIDRA e gera todos os parâmetros
    necessários para os gráficos e para o texto do relatório.
    """
    df_sidra = extrair_dados_completos_sidra(periodo=periodo)
    meses_disponiveis = df_sidra['mes_nome'].unique()
    
    # Mapeamento de abreviações dos meses para o eixo X do gráfico
    mapa_meses_abrev = {
        'janeiro': 'Jan', 'fevereiro': 'Fev', 'março': 'Mar', 'abril': 'Abr',
        'maio': 'Mai', 'junho': 'Jun', 'julho': 'Jul', 'agosto': 'Ago',
        'setembro': 'Set', 'outubro': 'Out', 'novembro': 'Nov', 'dezembro': 'Dez'
    }

    nomes_meses_formatados = []
    series_historicas = {estrato: [] for estrato in ESTRATOS_NOMES}
    ipca_geral_historico = []

    for mes in meses_disponiveis:
        # Extrai o nome do mês abreviado
        mes_puro = mes.split()[0].lower()
        nomes_meses_formatados.append(mapa_meses_abrev.get(mes_puro, mes_puro[:3].capitalize()))

        df_mes = df_sidra[df_sidra['mes_nome'] == mes]
        
        # IPCA Geral do mês
        ipca_geral_linha = df_mes[df_mes['codigo_item'] == '7169']
        ipca_val = ipca_geral_linha['variacao_mensal'].values[0] if not ipca_geral_linha.empty else 0.0
        ipca_geral_historico.append(ipca_val)

        # Mapeia as variações por código numérico exato
        variacoes_dict = {}
        for cod_sidra, nome_cat in CODIGOS_SUBGRUPOS_SIDRA.items():
            linha = df_mes[df_mes['codigo_item'] == cod_sidra]
            variacoes_dict[nome_cat] = linha['variacao_mensal'].values[0] if not linha.empty else 0.0
            
        # Categoria residual (média ponderada dos demais)
        variacoes_dict['Outros alimentos'] = np.mean(list(variacoes_dict.values()))

        # Cálculo da inflação ponderada para cada uma das 6 faixas
        for idx_estrato, estrato in enumerate(ESTRATOS_NOMES):
            inflacao_ponderada = sum(
                PESOS_POF_NORMALIZADOS[cat][idx_estrato] * variacoes_dict[cat]
                for cat in PESOS_POF_NORMALIZADOS
            )
            series_historicas[estrato].append(round(inflacao_ponderada, 2))

    # Mês mais recente
    mes_atual_nome = meses_disponiveis[-1]
    df_mes_atual = df_sidra[df_sidra['mes_nome'] == mes_atual_nome]

    # Decomposição detalhada dos subitens para o mês atual
    decomposicao_itens = []
    for cat in PESOS_POF_NORMALIZADOS:
        peso_baixa = PESOS_POF_NORMALIZADOS[cat][0]
        peso_alta = PESOS_POF_NORMALIZADOS[cat][5]
        
        # Encontra variação do item
        var_item = 0.0
        for cod_sidra, nome_cat in CODIGOS_SUBGRUPOS_SIDRA.items():
            if nome_cat == cat:
                linha = df_mes_atual[df_mes_atual['codigo_item'] == cod_sidra]
                if not linha.empty:
                    var_item = linha['variacao_mensal'].values[0]
                break

        contrib_marginal = (peso_baixa - peso_alta) * var_item
        decomposicao_itens.append({
            'categoria': cat,
            'peso_baixa': round(peso_baixa * 100, 1),
            'peso_alta': round(peso_alta * 100, 1),
            'variacao_mes': round(var_item, 2),
            'contrib_marginal': round(contrib_marginal, 3)
        })

    df_decomposicao = pd.DataFrame(decomposicao_itens).sort_values(by='contrib_marginal', ascending=False)

    return {
        'mes_atual': mes_atual_nome.capitalize(),
        'meses_eixo_x': nomes_meses_formatados,
        'series_historicas': series_historicas,
        'ipca_geral_mes': f"{ipca_geral_historico[-1]:+.2f}".replace('.', ','),
        'decomposicao': df_decomposicao,
        'variacoes_mes_atual': [series_historicas[e][-1] for e in ESTRATOS_NOMES]
    }

if __name__ == "__main__":
    resultado = processar_calculos_relatorio(periodo="last 6")
    print(f"\nSucesso! Mês de referência processado: {resultado['mes_atual']}")
    print(f"IPCA Geral do Mês: {resultado['ipca_geral_mes']}%")
    print(resultado['decomposicao'].head(5))
```
