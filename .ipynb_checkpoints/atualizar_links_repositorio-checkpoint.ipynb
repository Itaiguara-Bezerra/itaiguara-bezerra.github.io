```python
# ==============================================================================
# PROJETO: MONITOR DE INFLAÇÃO DE ALIMENTAÇÃO DOMICILIAR POR ESTRATO DE RENDA
# SCRIPT DE ATUALIZAÇÃO AUTOMÁTICA DOS LINKS DE REPOSITÓRIO E NOTA TÉCNICA
# Autor: Itaiguara de Oliveira Bezerra | Economista Sênior e de Dados
# ==============================================================================

import os
import re
import sys


def extrair_id_google_drive(url_ou_id: str) -> str:
  """Extrai o ID alfanumérico único de um arquivo do Google Drive,

  seja fornecida a URL completa de compartilhamento ou apenas o ID.
  """
  padrao = r"(?:[-\w]{25,})|(?:drive\.google\.com\/(?:file\/d\/|open\?id=)([-\w]+))"
  match = re.search(r"[-\w]{25,}", url_ou_id)
  if match:
    return match.group(0)
  return url_ou_id.strip()


def gerar_link_download_direto(id_arquivo: str) -> str:
  """Formata a URL para forçar o download automático pelo navegador."""
  id_limpo = extrair_id_google_drive(id_arquivo)
  return f"https://drive.google.com/uc?export=download&id={id_limpo}"


def atualizar_tex_relatorio(caminho_tex_relatorio: str, link_pdf_nt: str):
  """Substitui ou insere a chamada com o link direto no LaTeX do relatório."""
  if not os.path.exists(caminho_tex_relatorio):
    raise FileNotFoundError(
        f"Arquivo LaTeX não encontrado: {caminho_tex_relatorio}"
    )

  with open(caminho_tex_relatorio, "r", encoding="utf-8") as f:
    conteudo = f.read()

  # Expressão que localiza a linha da Nota Técnica Metodológica no LaTeX
  padrao_busca = (
      r"(\\textbf\{Nota Técnica Metodológica \(PDF\):\}.*?)"
      r"(\\href\{[^\}]*\}\{[^\}]*\}|download direto)"
  )

  nova_linha = (
      r"\1\\href{" + link_pdf_nt + r"}{\textbf{\underline{download direto}}}"
  )

  novo_conteudo, total_subs = re.subn(
      padrao_busca, nova_linha, conteudo, flags=re.IGNORECASE
  )

  if total_subs == 0:
    # Caso o texto original seja apenas "download direto" sem o href anterior
    padrao_simples = r"Nota Técnica Metodológica \(PDF\):(\s*)download direto"
    nova_chamada = (
        r"Nota Técnica Metodológica (PDF):\1\\href{"
        + link_pdf_nt
        + r"}{\textbf{\underline{download direto}}}"
    )
    novo_conteudo, total_subs = re.subn(
        padrao_simples, nova_chamada, conteudo, flags=re.IGNORECASE
    )

  with open(caminho_tex_relatorio, "w", encoding="utf-8") as f:
    f.write(novo_conteudo)

  print(
      f"[OK] {total_subs} ocorrência(s) atualizada(s) no arquivo"
      f" '{caminho_tex_relatorio}'."
  )
  print(f"[LINK APLICADO]: {link_pdf_nt}")


# ==============================================================================
# EXECUÇÃO PRINCIPAL
# ==============================================================================
if __name__ == "__main__":
  # 1. Caminho do arquivo .tex do seu Relatório de 5 páginas
  ARQUIVO_TEX_RELATORIO = "relatorio_mensal_ipca_renda.tex"

  # 2. COLE AQUI O LINK DE COMPARTILHAMENTO OU O ID DA NOTA TÉCNICA DO GOOGLE DRIVE:
  # Exemplo de link completo ou só o ID longo:
  # LINK_OU_ID = "https://drive.google.com/file/d/1a2b3c4d5e6f7g8h9i0j/view?usp=sharing"
  LINK_OU_ID = "1a2b3c4d5e6f7g8h9i0j_SEU_ID_AQUI"

  # Monta o link direto de download
  url_download_direto = gerar_link_download_direto(LINK_OU_ID)

  print("=" * 80)
  print("ATUALIZADOR DE REPOSITÓRIO: NOTA TÉCNICA NO GOOGLE DRIVE")
  print("=" * 80)

  if "SEU_ID_AQUI" in url_download_direto:
    print(
        "[AVISO] Substitua 'LINK_OU_ID' no script pelo link ou ID real do PDF"
        " que você subiu no Google Drive."
    )
  else:
    atualizar_tex_relatorio(ARQUIVO_TEX_RELATORIO, url_download_direto)

  print("=" * 80)
```
