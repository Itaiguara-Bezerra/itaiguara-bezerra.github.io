```python
import os
import sys
import shutil
import argparse
import datetime
import pandas as pd

# ==============================================================================
# CONFIGURAÇÕES DE DOWNLOAD E PASTAS
# ==============================================================================
BASE_DIR = r"C:\Users\ITAIGUARA\Dropbox\Relatorio_IPCA_Domicilio_Renda"

# Preencha com as URLs de download direto (GitHub Releases/Pages ou Google Drive)
CONFIG_DOWNLOADS = {
    "inflacao_mensal_ipca_renda.parquet": {
        "url": "https://itaiguara-bezerra.github.io/dados/inflacao_mensal_ipca_renda.parquet",
        "colunas_obrigatorias": [
            "data_base",
            "Renda Muito Baixa",
            "Renda Baixa",
            "Renda Média-Baixa",
            "Renda Média",
            "Renda Média-Alta",
            "Renda Alta"
        ],
        "tamanho_min_kb": 2
    },
    "ipca_alimentacao_subitens_historico_completo.parquet": {
        "url": "https://itaiguara-bezerra.github.io/dados/ipca_alimentacao_subitens_historico_completo.parquet",
        "colunas_obrigatorias": ["data_base", "subitem_nome", "variacao_preco_t"],
        "tamanho_min_kb": 10
    },
    "pesos_pof_normalizados.parquet": {
        "url": "https://itaiguara-bezerra.github.io/dados/pesos_pof_normalizados.parquet",
        "colunas_obrigatorias": ["codigo_subitem", "Renda Muito Baixa", "Renda Alta"],
        "tamanho_min_kb": 2
    }
}

DIR_TMP = os.path.join(BASE_DIR, ".tmp_download_parquet")
DIR_BACKUP = os.path.join(BASE_DIR, ".backup_parquet")


def baixar_arquivo(url, destino):
    """Baixa o arquivo via requests (HTTP direto) ou via gdown caso seja link do Google Drive."""
    if "drive.google.com" in url:
        try:
            import gdown
        except ImportError:
            raise ImportError("Para baixar links do Google Drive é necessário ter a biblioteca 'gdown' instalada (pip install gdown).")
        saida = gdown.download(url, destino, quiet=False, fuzzy=True)
        if not saida or not os.path.exists(destino):
            raise ConnectionError(f"Falha ao realizar download via gdown da URL: {url}")
    else:
        import requests
        resposta = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, stream=True, timeout=90)
        if resposta.status_code != 200:
            raise ConnectionError(f"Falha ao baixar {url} - Status Code HTTP: {resposta.status_code}")
        with open(destino, "wb") as f:
            for bloco in resposta.iter_content(chunk_size=8192):
                if bloco:
                    f.write(bloco)


def validar_parquet(caminho, colunas_esperadas, tamanho_min_kb):
    """Verifica tamanho em disco e conformidade das colunas obrigatórias."""
    if not os.path.exists(caminho):
        return False, "Arquivo temporário não foi gravado no disco."
    
    tamanho_kb = os.path.getsize(caminho) / 1024.0
    if tamanho_kb < tamanho_min_kb:
        return False, f"Arquivo corrompido ou incompleto: tamanho {tamanho_kb:.2f} KB (esperado mínimo de {tamanho_min_kb} KB)."

    try:
        df = pd.read_parquet(caminho)
    except Exception as e:
        return False, f"Arquivo não é um Parquet legível ou está corrompido: {e}"

    if df.empty:
        return False, "O arquivo Parquet está vazio (0 linhas)."

    colunas_presentes = set(df.columns)
    faltantes = [c for c in colunas_esperadas if c not in colunas_presentes]
    if faltantes:
        return False, f"Schema divergente! Colunas ausentes no arquivo: {faltantes}"

    return True, f"OK ({len(df)} registros, {tamanho_kb:.1f} KB)"


def executar_pipeline(dry_run=False):
    print("=" * 70)
    print(f"PIPELINE DE DOWNLOAD E ATUALIZAÇÃO DE BASES | DRY-RUN: {dry_run}")
    print(f"Diretório Base: {BASE_DIR}")
    print("=" * 70)

    os.makedirs(DIR_TMP, exist_ok=True)
    os.makedirs(DIR_BACKUP, exist_ok=True)

    timestamp_backup = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    arquivos_validados = []

    try:
        for nome_arquivo, cfg in CONFIG_DOWNLOADS.items():
            caminho_tmp = os.path.join(DIR_TMP, nome_arquivo)
            print(f"\n[1/3 Baixando] -> {nome_arquivo}...")
            
            try:
                baixar_arquivo(cfg["url"], caminho_tmp)
            except Exception as erro_dl:
                print(f"[ERRO DE DOWNLOAD] {erro_dl}")
                raise erro_dl

            print(f"[2/3 Validando] -> {nome_arquivo}...")
            valido, mensagem = validar_parquet(caminho_tmp, cfg["colunas_obrigatorias"], cfg["tamanho_min_kb"])
            
            if not valido:
                print(f"[REJEITADO] {nome_arquivo}: {mensagem}")
                raise ValueError(f"Validação falhou para {nome_arquivo}: {mensagem}")
            
            print(f"[VALIDADO COM SUCESSO] {mensagem}")
            arquivos_validados.append(nome_arquivo)

        if dry_run:
            print("\n" + "=" * 70)
            print("MODO DRY-RUN CONCLUÍDO: Todos os arquivos foram baixados e validados com sucesso.")
            print("Nenhum arquivo de produção foi modificado.")
            print("=" * 70)
            return

        print("\n[3/3 Atualização Atômica e Backup]")
        for nome_arquivo in arquivos_validados:
            caminho_origem_tmp = os.path.join(DIR_TMP, nome_arquivo)
            caminho_destino_oficial = os.path.join(BASE_DIR, nome_arquivo)

            # Backup da versão existente
            if os.path.exists(caminho_destino_oficial):
                nome_backup = f"{nome_arquivo.replace('.parquet', '')}_{timestamp_backup}.parquet"
                caminho_backup = os.path.join(DIR_BACKUP, nome_backup)
                shutil.copy2(caminho_destino_oficial, caminho_backup)
                print(f" - Backup criado: {nome_backup}")

            # Substituição atômica no mesmo sistema de arquivos
            os.replace(caminho_origem_tmp, caminho_destino_oficial)
            print(f" - Arquivo oficial atualizado: {nome_arquivo}")

        print("\n" + "=" * 70)
        print("BASE DE DADOS ATUALIZADA COM SUCESSO E SEM RISCO DE CORRUPÇÃO!")
        print("=" * 70)

    finally:
        if os.path.exists(DIR_TMP):
            shutil.rmtree(DIR_TMP, ignore_errors=True)


# ==============================================================================
# EXECUÇÃO COMPATÍVEL COM JUPYTER NOTEBOOK (.ipynb) E LINHA DE COMANDO (.py)
# ==============================================================================
parser = argparse.ArgumentParser(
    description="Download seguro e validação atômica das bases Parquet."
)
parser.add_argument(
    "--dry-run",
    action="store_true",
    help="Baixa e valida sem substituir os arquivos de produção."
)

# parse_known_args ignora os parâmetros internos que o Jupyter Notebook injeta
args, _ = parser.parse_known_args()

# Se quiser rodar apenas teste no notebook sem substituir nada, mude para: executar_pipeline(dry_run=True)
executar_pipeline(dry_run=args.dry_run)
```
